```python
def leap_year(yr):
  if yr/4 == int(yr/4):
    if yr/100 == int(yr/100): # century years are not leap years as per Gregorian calendar
      if yr/400 == int(yr/400): # except those century years, which are divisible by 400
        return True
      else:
        return False
    else:
      return True
  else:
    return False
yr = 600
if leap_year(yr):
  print(f"{yr} is a leap year")
else:
  print(f"{yr} is not a leap year")
```

```python
# 7. Find second largest number in a list without sorting
```

```python
#Method-1
def second_largest(numbers):
  if len(numbers)<2:
    return "List must have atleast two numbers"
  largest = max(numbers)
  numbers.remove(largest)
  second_largest = max(numbers)
  return second_largest

my_list = [10, 5, 8, 20, 15]
result = second_largest(my_list)
print("The second largest number is: ",result)

#Method-2
def second_largest(numbers):
  if len(numbers)<2:
    return "List must have atleast two elements"
  max1 = max2 = None

  for num in numbers:
    if max1 is None or num > max1:
      max2 = max1
      max1 = num
    elif num != max1 and (max2 is None or num > max2):
      max2 = num
  if max2 is None:
    return "There is no second largest element."
  return max2

my_list = [10, 5, 8, 20, 15]
result = second_largest(my_list)
print("The second largest number is: ", result)
```

```python
#12. Divide two integers without using multiplication, division and mod operator and any builtin Methods
```

```python
def divide(dividend, divisor):
    if divisor == 0:
        raise ValueError("Division by zero")

    quotient = 0

    while dividend >= divisor:
        dividend -= divisor
        quotient += 1

    return quotient


print(divide(10, 3))  # 3
print(divide(20, 5))  # 4
print(divide(15, 4))  # 3
```

```python
#15. Write a program that takes a string as input and returns its middle terms (Don't use built-in methods)
```

```python
def middle_terms(text):
    count = 0

    # Find length manually
    for _ in text:
        count += 1

    if count % 2 == 1:
        # Odd length -> one middle character
        mid = count // 2

        index = 0
        for ch in text:
            if index == mid:
                return ch
            index += 1

    else:
        # Even length -> two middle characters
        mid1 = count // 2 - 1
        mid2 = count // 2

        result = ""
        index = 0

        for ch in text:
            if index == mid1 or index == mid2:
                result += ch
            index += 1

        return result


print(middle_terms("hello"))      # l
print(middle_terms("python"))     # th
print(middle_terms("abcde"))      # c
print(middle_terms("abcd"))       # bc
```

```python
#19. Write a function to check if two strings are Anagrams without using built-in methods or sorting
```

```python
def is_anagram(s1, s2):

    # Calculate lengths manually
    len1 = 0
    for _ in s1:
        len1 += 1

    len2 = 0
    for _ in s2:
        len2 += 1

    if len1 != len2:
        return False

    freq = {}

    # Count characters in first string
    for ch in s1:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1

    # Decrease count using second string
    for ch in s2:
        if ch not in freq:
            return False

        freq[ch] -= 1

        if freq[ch] < 0:
            return False

    # Verify all counts are zero
    for key in freq:
        if freq[key] != 0:
            return False

    return True


print(is_anagram("listen", "silent"))   # True
print(is_anagram("earth", "heart"))     # True
print(is_anagram("hello", "world"))     # False
```

```python
def print_from_middle(word):

    n = 0
    for _ in word:
        n += 1

    mid = n // 2

    left = mid
    right = mid

    while left >= 0 and right < n:
        result = ""

        i = left
        while i <= right:
            result += word[i]
            i += 1

        print(result)

        left -= 1
        right += 1

print_from_middle("PROGRAM")
```

```python
#22. Generate Parentheses
# Given n pairs of parentheses, write a function to generate all combinations of well-formed parentheses.
#Input: n=3
#output: ["((()))", "(()())","(())()", "()(())","()()()"]
```

```python
def generate_parentheses(n):

    result = []

    def backtrack(current, open_count, close_count):

        if len(current) == 2 * n:
            result.append(current)
            return

        # Add opening parenthesis
        if open_count < n:
            backtrack(current + "(", open_count + 1, close_count)

        # Add closing parenthesis
        if close_count < open_count:
            backtrack(current + ")", open_count, close_count + 1)

    backtrack("", 0, 0)

    return result


print(generate_parentheses(4))
```

```python
#23. Write a program to give the following output for the given input
# Input: a1b10, Output: abbbbbbbbbb
# Input: b3c2d1, Output: bbbccd
```

```python
def expand_string(s):

    result = ""
    i = 0

    while i < len(s):

        ch = s[i]
        i += 1

        num = 0

        while i < len(s) and '0' <= s[i] <= '9':
            num = num * 10 + (ord(s[i]) - ord('0'))
            i += 1

        result += ch * num

    return result


print(expand_string("a1b10"))
print(expand_string("b3c2d1"))
```

```python
#25. Remove unbalanced parentheses in a given expression
# Input: ((abc)((de)) Output: ((abc)(de))
# Input: (a(b)))(c(d) Output: (a(b))(cd)
# Input: (((ab) Output: (ab)
```

```python
def balance_parentheses(s):

    stack = []
    keep = [False] * len(s)

    for i in range(len(s)):
        if s[i] == '(':
            stack.append(i)

        elif s[i] == ')':
            if stack:
                open_idx = stack.pop()

                keep[open_idx] = True
                keep[i] = True

    result = ""

    for i in range(len(s)):
        if s[i] not in "()":
            result += s[i]
        elif keep[i]:
            result += s[i]

    return result


print(balance_parentheses("((abc)((de))"))
print(balance_parentheses("(a(b)))(c(d)"))
print(balance_parentheses("(((ab)"))
```

```python
# 26. Given an integer N, the task is to convert the given number into words
# Input: N = 437 Output: Four Hundred and Thirty Seven
```

```python
def two_digit_to_words(n):
    ones = ["", "One", "Two", "Three", "Four", "Five",
            "Six", "Seven", "Eight", "Nine"]

    teens = ["Ten", "Eleven", "Twelve", "Thirteen", "Fourteen",
             "Fifteen", "Sixteen", "Seventeen", "Eighteen", "Nineteen"]

    tens = ["", "", "Twenty", "Thirty", "Forty",
            "Fifty", "Sixty", "Seventy", "Eighty", "Ninety"]

    if n < 10:
        return ones[n]

    elif n < 20:
        return teens[n - 10]

    else:
        result = tens[n // 10]

        if n % 10:
            result += " " + ones[n % 10]

        return result


def three_digit_to_words(n):
    result = ""

    if n >= 100:
        result += two_digit_to_words(n // 100) + " Hundred"

        if n % 100:
            result += " and "

    result += two_digit_to_words(n % 100)

    return result


def number_to_words(n):

    if n == 0:
        return "Zero"

    result = ""

    crore = n // 10000000
    n %= 10000000

    lakh = n // 100000
    n %= 100000

    thousand = n // 1000
    n %= 1000

    if crore:
        result += two_digit_to_words(crore) + " Crore "

    if lakh:
        result += two_digit_to_words(lakh) + " Lakh "

    if thousand:
        result += two_digit_to_words(thousand) + " Thousand "

    if n:
        result += three_digit_to_words(n)

    return result.strip()


# Test Cases
print(number_to_words(437))
print(number_to_words(43907))
print(number_to_words(1234567))
print(number_to_words(98765432))
```

```python
def number_to_excel_column(n):
    result = ""

    while n > 0:
        n -= 1                # Adjust for 1-based indexing
        remainder = n % 26

        result = chr(65 + remainder) + result

        n //= 26

    return result


# Test Cases
print(number_to_excel_column(1))
print(number_to_excel_column(26))
print(number_to_excel_column(27))
print(number_to_excel_column(1000))
```

```python
#28. Write a program to print the following output for the given input, you can assume the string is of odd length.
# Input 12345,
# Output:
# 1       5
#   2   4
#     3
#  2    4
# 1       5
```

```python
def print_pattern(s):

    n = len(s)
    mid = n // 2

    # Upper half including middle
    for i in range(mid + 1):

        # Leading spaces
        for _ in range(i):
            print(" ", end="")

        print(s[i], end="")

        if i != mid:
            spaces = (n - 2 * i - 2) * 2 + 1

            for _ in range(spaces):
                print(" ", end="")

            print(s[n - 1 - i], end="")

        print()

    # Lower half
    for i in range(mid - 1, -1, -1):

        for _ in range(i):
            print(" ", end="")

        print(s[i], end="")

        spaces = (n - 2 * i - 2) * 2 + 1

        for _ in range(spaces):
            print(" ", end="")

        print(s[n - 1 - i], end="")

        print()


print_pattern("Welcome")
```

```python
#29. Given an integer, the task is to write a program to convert integer to roman.
# Input: num=3749, Output: "MMMDCCXLIX"
# Input: num=58, Output: "LVIII"
# Input: num=1994, Output: "MCMXCIV"
```

```python
def int_to_roman(num):

    values = [
        1000, 900, 500, 400,
        100, 90, 50, 40,
        10, 9, 5, 4, 1
    ]

    romans = [
        "M", "CM", "D", "CD",
        "C", "XC", "L", "XL",
        "X", "IX", "V", "IV", "I"
    ]

    result = ""

    for i in range(len(values)):

        while num >= values[i]:
            result += romans[i]
            num -= values[i]

    return result


print(int_to_roman(3749))
print(int_to_roman(58))
print(int_to_roman(1994))
```
