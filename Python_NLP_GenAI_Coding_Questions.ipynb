4. Write a Python function to generate first n terms of the fibonacci sequence. The fibonacci sequence is a series of numbers where each number is the sum of the two preceding ones, usually starting with 0 and 1

```python
def fibonacci(n):
  fib_seq = [0, 1] # initialize first 2 terms of fib_seq
  if n <= 0:
    return []
  while len(fib_seq) < n:
    next_term = fib_seq[-1] + fib_seq[-2]
    fib_seq.append(next_term)
  return fib_seq
print(fibonacci(8))
```

5. Write a function that checks whether a given number is a prime number. A prime number is a positive integer greater than 1 that has no positive integer divisors other than 1 and itself

```python
def prime_number(num):
  flag = 0
  if num <= 0:
    print("Negative integers are not allowed")
    return -1
  elif num == 1:
    flag = 1
  else:
    for i in range(2, num):
      if num % i == 0:
        flag = 1
        break

  if flag:
    print("Not a Prime Number")
  else:
    print("Prime Number")

prime_number(0)
prime_number(1)
prime_number(2)
prime_number(3)
```

```python
def anagram(word1, word2):
  word1 = word1.replace(" ","").lower()
  word2 = word2.replace(" ","").lower()
  if sorted(word1) == sorted(word2):
    print(f"{word1} and {word2} are anagrams")
  else:
    print(f"{word1} and {word2} are not anagrams")
anagram('keen', 'knee')
anagram('listen', 'silent')
anagram('ahft', 'ahtt')
```

10. Find the second largest element in a list

```python
def second_largest(lst):
  max1 = lst[0]
  max2 = lst[1]
  if max1 < max2:
    f = max1
    max1 = max2
    max2 = f

  for i in lst:
    if i > max1:
      max2 = max1
      max1 = i
    elif i < max1 and max2 < i:
      max2 = i
  return max2

print(second_largest([23, 5,67, 98, 2845, 4657]))
print(second_largest([2, 5, 7, 897]))
```

12. Sort a list of numbers in ascending or descending order

```python
# Ascending
def lst_sort(lst):
  for i in range(len(lst)):
    for j in range(i+1, len(lst)):
      if lst[i] > lst[j]:
        lst[i], lst[j] = lst[j], lst[i]
  return lst

print(lst_sort([3, 67, 54, 3, 233, 76]))
```

```python
# Descending
def lst_sort(lst):
  for i in range(len(lst)):
    for j in range(i+1, len(lst)):
      if lst[i] < lst[j]:
        lst[i], lst[j] = lst[j], lst[i]
  return lst

print(lst_sort([3, 67, 54, 3, 233, 76]))
```

18. Armstrong - An Armstrong number (or narcissistic number) is a positive integer that equals the sum of its own digits, where each digit is raised to the power of the total number of digits in the number.

```python
def armstrong(num):
  no_digits = len(str(num))
  arms = 0
  for i in str(num):
    arms += (int(i) ** no_digits)
  return arms
num = 1534
out = armstrong(num)
if num == out:
  print("It is a armstrong number")
else:
  print("It's not a armstrong number")
```

```python
def min_max_digit(num):
  num = abs(num)

  digit = num % 10
  maximum = minimum = digit

  while num > 0:
    digit = num % 10
    if maximum < digit:
      maximum = digit
    elif minimum > digit:
      minimum = digit
    num //= 10
  return maximum, minimum

print(min_max_digit(26789))
```

23. Check Perfect Number - A perfect number is a positive integer that equals the sum of all its proper divisors, excluding the number itself. For example, \(6\) is a perfect number because its proper divisors are \(1, 2,\) and \(3\), and \(1 + 2 + 3 = 6\)

```python
def is_perfect(num):
  total = 0
  for i in range(1, num):
    if num % i == 0:
      total += i
  return total == num

num = 24
if is_perfect(num):
  print("It's a Perfect Number")
else:
  print("It's not a Perfect Number")
```

28. Find the indices of two numbers whose sum equals the target

```python
nums = [2, 7, 11, 15]
target = 9

def two_sum(nums, target):
    seen = {}

    for i, num in enumerate(nums):
        diff = target - num

        if diff in seen:
            return [seen[diff], i]

        seen[num] = i

    return []


print(two_sum(nums, target))
```

32. Retrieve Top-K Relevant Results

```python
# Method 1: Keyword-Based Similarity (TF-IDF + Cosine Similarity)

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Python is a popular programming language for data science.",
    "The weather today is sunny and beautiful.",
    "Machine learning and artificial intelligence use Python heavily.",
    "Exploring deep learning models with neural networks in Python.",
    "Healthy recipes often include fresh vegetables and fruits."
]
query = "Looking for machine learning tools in Python."
K = 2

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
query_vector = vectorizer.transform([query])

similarity_scores = cosine_similarity(query_vector, tfidf_matrix).flatten() # `.flatten()` turns the resulting 2D array into a 1D array of scores

top_k_indices = np.argsort(similarity_scores)[-K:][::-1] # argsort sorts ascending; [-K:] gets the highest elements; [::-1] reverses it

print(f"Query: '{query}'\n")
print(f"Top-{K} Most Similar Documents:")
for rank, idx in enumerate(top_k_indices, 1):
    print(f"Rank {rank}: Score = {similarity_scores[idx]:.4f}")
    print(f"Document: {documents[idx]}\n")
```

```python
# Method 2: Semantic Similarity (Sentence Transformers)
import numpy as np
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')

documents = [
    "Python is a popular programming language for data science.",
    "The weather today is sunny and beautiful.",
    "Machine learning and artificial intelligence use Python heavily.",
    "Exploring deep learning models with neural networks in Python.",
    "Healthy recipes often include fresh vegetables and fruits."
]
query = "How can I build neural networks?"
K = 2

doc_embeddings = model.encode(documents, convert_to_tensor=True)
query_embedding = model.encode(query, convert_to_tensor=True)

similarity_scores = util.cos_sim(query_embedding, doc_embeddings).flatten()

scores_npc = similarity_scores.cpu().numpy()
top_k_indices = np.argsort(scores_npc)[-K:][::-1] # Convert tensor to numpy array for sorting

print(f"Query: '{query}'\n")
print(f"Top-{K} Semantically Similar Documents:")
for rank, idx in enumerate(top_k_indices, 1):
    print(f"Rank {rank}: Score = {scores_npc[idx]:.4f}")
    print(f"Document: {documents[idx]}\n")
```

36. Decorator Usage

Decorators are flexible way to modify or extend behavior of functions or methods, without changing their actual code.

A decorator is essentially a function that takes another function as an argument and returns a new function with enhanced functionality.
They are often used in scenarios such as logging, authentication and memoization, allowing us to add additional functionality to existing functions or methods in a clean, reusable way.

```python
# Decorators often need to work with functions that have arguments. We use *args and **kwargs so our wrapper can accept any number of arguments.

def decorator(add):
  def wrapper_func(*args, **kwargs):
    print("Before Decorator Function Call")
    print(add(*args, **kwargs))
    print("After Decorator Function Call")
    return
  return wrapper_func
@decorator
def add(a, b, c):
  return a + b + c

add(3, 4, 3)
```

**RECURSION** - Recursion is a programming technique where a function calls itself either directly or indirectly to solve a problem. A recursive function calls itself in its body.

```python
# Recursion-1 : Factorial of a number
def fact(num):
  if num == 0 or num == 1:
    return 1
  return num * fact(num-1)
print(fact(6))
```

```python
# Recursion-2 : Fibonacci
def fibb(num):
  if num <= 1:
    return num
  return fibb(num-1) + fibb(num-2)
print(fibb(7))
```

Why this is inefficient?
It recalculates the same values repeatedly, resulting in exponential time complexity. Memoization or dynamic programming improves it to linear time.

```python
# Recursion-4 : Sum of Digits
def sum_digits(num):
  if num  == 0:
    return 0
  return num % 10 + sum_digits(num // 10)
sum_digits(12345)
```

```python
# Decorator-1 without Arguments
def logger(func):
  def wrapper():
    print("Before Function call")
    func()
    print("After Function call")
  return wrapper

@logger
def greet():
  print("hello")

greet()
```

```python
# Decorator-2 With Arguments
def logger(add):
  def wrapper(a, b):
    print("Before Addition")
    add(a, b)
    print("After Addition")
  return wrapper

@logger
def add(a, b):
  print("Addition: ", a+b)

add(1, 2)
```

## Split Documents into Chunks

Why overlap?

Why 500?

When would you use semantic chunking instead?

```python
# Split a document into chunks of 500 characters with an overlap of 100.
from langchain_text_splitters import RecursiveCharacterTextSplitter
text = "To create a tuple in Python, you place comma-separated values inside parentheses, but because tuples are immutable, you cannot directly insert or append values to them after creation.\n\n Instead, you must use specific workarounds—such as concatenation or list conversion—to add elements and generate a new tuple. \n That's all about it!"
splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10
)

chunks = splitter.split_text(text)
print(chunks)

print(len(chunks))
```

**4. Cosine Similarity**

Why cosine similarity instead of Euclidean distance?

Answer: Cosine similarity measures the angle between vectors, making it less sensitive to vector magnitude and more suitable for semantic similarity.

```python
# Find the similarity between two sentences.
from sklearn.metrics.pairwise import cosine_similarity

emb1 = model.encode(["AI is amazing"])
emb2 = model.encode(["Artificial Intelligence is powerful"])

score = cosine_similarity(emb1, emb2)

print(score)
```
